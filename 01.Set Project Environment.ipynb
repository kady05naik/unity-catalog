### 2. Create External location

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `gizmobox-ext`
    URL 'abfss://gizmobox@databricksprostoageacc.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-course-ext-ac`)
    COMMENT 'External location for gizmobox'
```

### 3. Create Catalog databricks

```python
%sql
CREATE CATALOG IF NOT EXISTS gizmobox
    MANAGED LOCATION 'abfss://gizmobox@databricksprostoageacc.dfs.core.windows.net/'
    COMMENT 'THis is catalog for gizmobox data lakehouse'
```

### 4. Create Schemas:
1. Landing
2. bronze
3. silver
4. gold     

```python
%sql
SELECT current_catalog();
```

```python
%sql 
USE CATALOG gizmobox;

CREATE SCHEMA IF NOT EXISTS `landing`
    COMMENT 'This is schema for landing zone'
    MANAGED LOCATION 'abfss://gizmobox@databricksprostoageacc.dfs.core.windows.net/landing';

CREATE SCHEMA IF NOT EXISTS `bronze`
    COMMENT 'This is schema for bronze'
    MANAGED LOCATION 'abfss://gizmobox@databricksprostoageacc.dfs.core.windows.net/bronze';

CREATE SCHEMA IF NOT EXISTS `silver`
    COMMENT 'This is schema for silver'
    MANAGED LOCATION 'abfss://gizmobox@databricksprostoageacc.dfs.core.windows.net/silver';

CREATE SCHEMA IF NOT EXISTS `gold`
    COMMENT 'THIS is chema for gold'
    MANAGED LOCATION 'abfss://gizmobox@databricksprostoageacc.dfs.core.windows.net/gold';
```

### 5. Create Volume

```python
%sql
USE CATALOG gizmobox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS `operational_data`
    LOCATION 'abfss://gizmobox@databricksprostoageacc.dfs.core.windows.net/landing/operational_data'
    COMMENT 'This is volume for operational data'
```
