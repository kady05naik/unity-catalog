### Extract Data from Customers JSON file
1. Query Single File
2. Query list of files using wildcard characters
3. Query all the files in a folder


```sql
%fs ls /Volumes/gizmobox/landing/operational_data/customers/
```

### 1. Query single JSON file

```sql
SELECT * FROM json.`/Volumes/gizmobox/landing/operational_data/customers/customers_2024_10.json`
```

### 2. Query list of files using wildcard characters

```sql
SELECT * FROM json.`/Volumes/gizmobox/landing/operational_data/customers/customers_2024*.json`
```

### 3. Query all the files in a folder

```sql
SELECT * FROM json.`/Volumes/gizmobox/landing/operational_data/customers/`;
```

### 4. Select _metadata

```sql
SELECT 
    _metadata,
    *
from json.`/Volumes/gizmobox/landing/operational_data/customers/`;
```

### 5. Selecting file_path from metadata

```sql
SELECT 
    _metadata.file_path AS file_path, 
    *
FROM json.`/Volumes/gizmobox/landing/operational_data/customers/`;
```

### 6. Register files in Unity Catalog using Views

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.view_customers AS
    SELECT 
        _metadata.file_path AS file_path,
        *
    FROM json.`/Volumes/gizmobox/landing/operational_data/customers/`;
```

```sql
SELECT * FROM gizmobox.bronze.view_customers;
```

### 7. Create temporary view

```sql
CREATE OR REPLACE TEMPORARY VIEW temp_customers AS
    SELECT 
        _metadata.file_path AS file_path,
        *
    FROM json.`/Volumes/gizmobox/landing/operational_data/customers/`;
```

```sql
SELECT * FROM temp_customers;
```

```sql
DESCRIBE EXTENDED temp_customers;
```

### 8. Create Global View

```sql
CREATE OR REPLACE GLOBAL TEMPORARY VIEW globTempView_customers AS
    SELECT 
        *,
        _metadata.file_path AS file_path
    FROM json.`/Volumes/gizmobox/landing/operational_data/customers/`;
```

```sql
SELECT * FROM global_temp.globTempView_customers;
```
