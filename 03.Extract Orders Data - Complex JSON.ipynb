### 1. Query Orders File using JSON Format

```sql
SELECT * FROM json.`/Volumes/gizmobox/landing/operational_data/orders`
```

As you can see the file has corrupt records due to missing '' comma and other reasons 
![image_1779787974683.png](./image_1779787974683.png "image_1779787974683.png")

### 2. Query Orders files using TEXT Format

As data is currupt so we will read data using text format, to prevent data loss

```sql
SELECT * FROM text.`/Volumes/gizmobox/landing/operational_data/orders`
```

### So now there is no data loss, when you do not trust the source data read it into text format for bronze layer and later transform it

![image_1779788436764.png](./image_1779788436764.png "image_1779788436764.png")

### 3. Create Orders view in Bronze Schema

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.view_orders AS
    SELECT * FROM text.`/Volumes/gizmobox/landing/operational_data/orders`;
```

```sql
SELECT * FROM gizmobox.bronze.view_orders;
```
