### Extract Data From Memberships- Image Files
1. Query Memberships File using binaryFile Format
2. Create Memberships View in Bronze Schema

### 1. Query Memberships folder with Binary File Format

```sql
%fs ls /Volumes/gizmobox/landing/operational_data/memberships
```

```sql
SELECT * FROM binaryFile.`/Volumes/gizmobox/landing/operational_data/memberships/*/*`;
```

### 2. Create Membership View in Bronze Schema

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.view_membership
AS
    SELECT * from binaryFile.`/Volumes/gizmobox/landing/operational_data/memberships/*/*`;  
```

```sql
SELECT * FROM gizmobox.bronze.view_membership;
```
