### 2. Use read_files functions to overcome the limitations

```sql
SELECT * 
FROM read_files ('/Volumes/gizmobox/landing/operational_data/addresses',
                format =>'csv',
                delimiter=>'\t',
                header => True);
```

read_files() is a built-in Databricks SQL function that allows you to read raw files directly from cloud storage. It works like a table function — meaning it returns rows just like a regular table.
![image_1779799500189.png](./image_1779799500189.png "image_1779799500189.png")

### 3. Create Addresses view in the Bronze Layer

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.view_addresses 
AS
    SELECT * 
    FROM read_files ('/Volumes/gizmobox/landing/operational_data/addresses',
                    format =>'csv',
                    delimiter=>'\t',
                    header => True);
```

```sql
SELECT * FROM gizmobox.bronze.view_addresses ;
```
