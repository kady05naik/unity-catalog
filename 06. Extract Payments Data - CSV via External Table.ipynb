### 1. List the files from Payment folder

### 2. Create External Table

```sql
CREATE TABLE IF NOT EXISTS gizmobox.bronze.payments
    (payment_id INTEGER, order_id INTEGER, payment_timestamp TIMESTAMP, payment_status INTEGER, payment_method STRING)
USING CSV
OPTIONS (
    header= True,
    delimiter=','
)
LOCATION 'abfss://gizmobox@databricksprostoageacc.dfs.core.windows.net/landing/external_data/payments';
```

```sql
SELECT * FROM gizmobox.bronze.payments;
```

### 3. Demonstrate the effect of Adding / Updating / Deleting files

WhenI delete a file from Payments Folder, 
`SELECT * FROM gizmobox.bronze.payments;` - Threw an error FAILED_READ_FILE.FILE_NOT_EXIST

### In this case you have to REFRESH the tablle- `REFRESH TABLE gizmobox.bronze.payments;`
After REFRESH it only provide the records of the existing files and not the deleted files

### 4. Demonstrate the effect of Droping the Table

When I used `DROP TABLE gizmobox.bronze.payments;`
It did not delete the files in External Location.
