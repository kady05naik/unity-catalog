### Extract Data - Using PySpark
1. Run SQL Commands using Python - spark.sql function
2. Spark Dataframe Reader API 
3. Read tables using spark.table function

### 1. Run SQL Commands using Python - spark.sql Function

```sql
%python
df=spark.sql('SELECT * FROM json.`/Volumes/gizmobox/landing/operational_data/customers`')
display(df)
```

### 2. Spark dataframeReader API


```sql
%python
df=spark.read.json('/Volumes/gizmobox/landing/operational_data/customers')
display(df)
```

### 3. Spark table function

```sql
%python
df=spark.table('gizmobox.bronze.view_customers')
display(df)
```

```sql
%python
df=spark.read.table('gizmobox.bronze.view_customers')
display(df)
```
